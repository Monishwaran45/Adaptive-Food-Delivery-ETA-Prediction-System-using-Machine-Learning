```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use("ggplot")

import warnings
warnings.filterwarnings("ignore")
```

```python
riders = pd.read_csv("../dataset/processed/riders_clean.csv")

print(riders.shape)

riders.head()
```

```python
print(riders.info())
```

```python
riders.describe(include="all")
```

```python
plt.figure(figsize=(10,5))

sns.histplot(
    riders["Time_taken (min)"],
    bins=30,
    kde=True
)

plt.title("Distribution of Delivery Time")
plt.xlabel("Delivery Time (Minutes)")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(
    riders["Delivery_person_Age"],
    bins=20,
    kde=True
)

plt.title("Delivery Person Age")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Delivery_person_Ratings",
    data=riders
)

plt.xticks(rotation=45)

plt.title("Delivery Ratings")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Road_traffic_density",
    data=riders
)

plt.title("Traffic Density Distribution")

plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    x="Weather_conditions",
    data=riders
)

plt.xticks(rotation=45)

plt.title("Weather Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Type_of_vehicle",
    data=riders
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Type_of_order",
    data=riders
)

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="City",
    data=riders
)

plt.show()
```

```python
plt.figure(figsize=(12,8))

numeric = riders.select_dtypes(include=np.number)

sns.heatmap(
    numeric.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Matrix")

plt.show()
```
