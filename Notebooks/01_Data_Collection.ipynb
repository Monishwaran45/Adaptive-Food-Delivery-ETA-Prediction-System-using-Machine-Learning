```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
DATASET_PATH = "../dataset"

orders = pd.read_csv(f"{DATASET_PATH}/Orders.csv")
restaurants = pd.read_csv(f"{DATASET_PATH}/Restaurants.csv")
riders = pd.read_csv(f"{DATASET_PATH}/riders.csv")
customers = pd.read_csv(f"{DATASET_PATH}/customer.csv")
weather = pd.read_csv(f"{DATASET_PATH}/weather.csv")
gps = pd.read_csv(f"{DATASET_PATH}/gps_tracking.csv")
```

```python
datasets = {
    "Orders": orders,
    "Restaurants": restaurants,
    "Riders": riders,
    "Customers": customers,
    "Weather": weather,
    "GPS": gps
}

for name, df in datasets.items():
    print("="*60)
    print(name)
    print(df.shape)
    print(df.info())
```

```python
for name, df in datasets.items():

    print("\n",name)

    print(df.isnull().sum())
```

```python
for name, df in datasets.items():

    print(name)

    print(df.duplicated().sum())
```

```python
for name, df in datasets.items():

    print(name)

    print(df.dtypes)
```
