```python
import pandas as pd
import numpy as np
```

```python
riders = pd.read_csv("../dataset/riders.csv")

print(riders.shape)
riders.head()
```

```python
missing = pd.DataFrame({
    "Missing": riders.isnull().sum(),
    "Percent": round(riders.isnull().mean()*100,2)
})

missing.sort_values("Percent", ascending=False)
```

```python
riders.replace(
    ["NaN", "nan", "NULL", "None", ""],
    np.nan,
    inplace=True
)
```

```python
num_cols = [
    "Delivery_person_Age",
    "Delivery_person_Ratings",
    "multiple_deliveries"
]

for col in num_cols:
    riders[col] = riders[col].fillna(riders[col].median())
```

```python
cat_cols = [
    "Weather_conditions",
    "Road_traffic_density",
    "Festival",
    "City"
]

for col in cat_cols:
    riders[col] = riders[col].fillna(riders[col].mode()[0])
```

```python
riders["Time_Orderd"] = riders["Time_Orderd"].fillna(
    riders["Time_Order_picked"]
)
```

```python
riders["Time_Orderd"] = riders["Time_Orderd"].astype(str).str.strip()

riders["Time_Order_picked"] = riders["Time_Order_picked"].astype(str).str.strip()
```

```python
riders["Order_Date"] = pd.to_datetime(
    riders["Order_Date"],
    format="%d-%m-%Y"
)
```

```python
riders = riders[
    riders["Delivery_person_Ratings"].between(1, 5)
]
```

```python
riders = riders[
    riders["Delivery_person_Age"].between(18, 60)
]
```

```python
print("Duplicates:", riders.duplicated().sum())

riders = riders.drop_duplicates()
```

```python
print(riders.shape)

print("\nMissing Values")
print(riders.isnull().sum())

print("\nData Types")
print(riders.dtypes)

print("\nSample Time Values")
print(riders[["Time_Orderd", "Time_Order_picked"]].head())
```

```python
riders.to_csv(
    "../dataset/processed/riders_clean.csv",
    index=False
)
```
